 Number Guessing Game

Problem Statement: Create a Python program that implements a simple number guessing game using a while loop.

The program should make use of control statements such as else, break, and continue.

```python

import random

# Generate a random number between 1 and 10
secret_number = random.randint(1, 10)

# Ask the user to guess the number
limit = 1

while limit <= 3:
    number = int(input("Guess a number between 1 and 10: "))

    if number < 1 or number > 10:
        print("Your guess is out of range. Please guess a number between 1 and 10.")
        continue

    limit += 1

    if number < secret_number:
        print("Too low. Try again.")

    elif number > secret_number:
        print("Too high. Try again.")

    elif number == secret_number:
        print("Congratulations! You guessed the correct number.")
        break

else:
    print("Better luck next time!")
    print("The correct number was:", secret_number)
```

### For Loop:

Multiplication Table Generator

Problem Statement: 
Create a Python program that generates and prints a
multiplication table (from 1 to 10) for a given number using a for loop and the range
function.

```python
#Asking the user to enter a number for which they want to generate a multiplication table.
Number=int(input("Enter any number for multiplication table:"))

print(f"\nMultiplication Table of {Number}:")

for i in range(1,11):  # specifying the table factors should from 1 to 1o
    Product=Number*i   # Operation done
    print(f"{Number}x{i}={Product}")
```
